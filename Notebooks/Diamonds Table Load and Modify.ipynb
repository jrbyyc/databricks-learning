```python
# Load diamonds dataset and print schema to confirm load ok
df = spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/Volumes/workspace/default/diamonds/diamonds.csv")

df.printSchema()
```

```python
# Save the DataFrame as a managed Delta table so SQL cells can query/modify it
(df.write
    .mode("overwrite")
    .saveAsTable("workspace.default.diamonds"))
```
